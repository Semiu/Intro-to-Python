```python
# R-1.1

def is_multiple(n, m):
    """ check n as a multiple of m """

    if n%m == 0:
        return True
    return False

assert is_multiple(10, 5) == True, "Multiple"

assert is_multiple(10, 3) == False, "Not a multiple"
# Uncomment the line below to see the assertion error
# assert is_multiple(12, 5) == True, "Built to fail because there are not Multiple"
```

```python
# R-1.2

def is_even(n):
    """ check n as an even number """

    if (n & 1) == 0:
        return True
    return False

assert is_even(10) == True, "Even"
assert is_even(11) == False, "Not even"
```

```python
# R-1.3
def minmax(data):

    if not data:
        raise ValueError("Provided data has no value")
    else:
        min_value = 0
        max_value = 0
        for ele in data:
            if ele > max_value:
                max_value = ele
            if ele < min_value:
                min_value = ele
                
        return min_value, max_value


print(minmax([0,-1,15,8,2,1]))
```

```python
# R-1.4

def sum_of_squares(n):
    
    square_sum = 0
    for i in range(1,n-1,1):
        square_sum += i*i

    return square_sum
        
print(sum_of_squares(8))
```

```python
# R-1.5
from math import pow
def sum_of_squares(n):
    return sum([pow(i,2) for i in range(1,n-1,1)])

print(sum_of_squares(8))
```
